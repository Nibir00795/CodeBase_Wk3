Task 2 builds a small student DataFrame with pandas. Task 6 draws it with Plotly.

```python
# Task 2 — student data, assembled by Md Jonayed Hossain Chowdhury
import pandas as pd

# Sample data. The last entry is my own row, added as the lab asks.
data = {
    'Name': ['Alice', 'Bob', 'Carol', 'David', 'Eve', 'Frank', 'Grace',
             'Hank', 'Ivy', 'Jack', 'Jonayed'],
    'Year': ['Sophomore', 'Junior', 'Senior', 'Freshman', 'Senior',
             'Junior', 'Sophomore', 'Senior', 'Freshman', 'Junior', 'Graduate'],
    'Major': ['IT', 'CPS', 'STEM', 'CPS', 'IT', 'STEM', 'CPS', 'STEM',
              'IT', 'CPS', 'CPS']
}

# Create DataFrame
df = pd.DataFrame(data)

# Display DataFrame
print(df)
```

## Data Visualization
Task 6. The same eleven rows, drawn with the chart from the handout. Year is a category, not a quantity, so putting it on the y axis gives the bars heights that mean nothing: Plotly orders the categories as Sophomore, Senior, Freshman, Junior, Graduate, which is neither academic order nor alphabetical. The x axis also groups the names by major rather than keeping them in table order. Both are artefacts of the encoding rather than facts about the students, which is what the next cell addresses.

```python
import plotly.express as px

# Create a bar plot
fig = px.bar(df, x='Name', y='Year', color='Major', title='Student data')
fig.show()
```

### A second view that answers a question
The chart above puts a category on the y axis, so its bar heights carry no magnitude. Counting students per major is the comparison the data actually supports, so here it is as a count.

```python
counts = df['Major'].value_counts().reset_index()
counts.columns = ['Major', 'Students']

fig2 = px.bar(counts, x='Major', y='Students',
              title='Students per major (n = %d)' % len(df),
              text='Students')
fig2.update_traces(marker_color='#2a78d6', textposition='outside')
fig2.update_layout(yaxis_title='Students', xaxis_title='Major',
                   plot_bgcolor='#fcfcfb', paper_bgcolor='#fcfcfb',
                   font_color='#52514e', showlegend=False)
fig2.update_yaxes(gridcolor='#eef0f2', dtick=1)
fig2.show()
```
